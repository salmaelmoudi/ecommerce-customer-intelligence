```python
import pandas as pd

print(pd.__version__)
```

```python
# Load the datasets
orders = pd.read_csv("../data/olist_orders_dataset.csv")
payments = pd.read_csv("../data/olist_order_payments_dataset.csv")
# Find delivered orders where the delivery date is missing
missing_delivery_date = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isna())
]
# Display the results
print("Number of delivered orders with missing delivery date:",
      len(missing_delivery_date))

print("\nOrders with missing delivery date:")
print(missing_delivery_date[
    ["order_id", "order_status", "order_purchase_timestamp",
     "order_delivered_customer_date"]
])
```

```python
# Challenge 1: Cross-Table Validation (Orphan Orders)
orphan_orders = orders[~orders['order_id'].isin(payments['order_id'])]
print(f"Orphan orders without payment records: {len(orphan_orders)}")

# Challenge 2: Value Validation (Zero/Negative Payments)
zero_or_neg_payments = payments[payments['payment_value'] <= 0]
print(f"Payments <= 0: {len(zero_or_neg_payments)}")

# Challenge 3: Date Validation (Time-traveling Deliveries)
# Ensure timestamps are parsed as datetime objects first
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])

time_travel_orders = orders[
    orders['order_delivered_customer_date'] < orders['order_purchase_timestamp']
]
print(f"Orders delivered before purchase timestamp: {len(time_travel_orders)}")
```
